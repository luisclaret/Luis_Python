# FOR LOOPS #

```python
for i in range(1, 20): # Aqui va de 1 a 19
    # Codigo
    print(i)

for i in range(20): # Aqui comienza desde 0 y va hasta el 19
    # codigo
    print(i)

for i in range(0, 20, 1):
    # codigo
    print(i)
```

```python
lista = [2, 4, 6, 8]

for i in lista:
    print(i * 10)

for i in range(len(lista)):
    print(lista[i] * 10)

for _ in range(10):
    print("Hola mundo")
```

```python
string = "tarea casa helado"

acumulador = ""

for i in string:
    if i == "h":
        break
    acumulador = acumulador + i

print(acumulador)
```

```python
string = "tarea casa helado"

acumulador = ""

for i in string:
    acumulador = i + acumulador

print(acumulador)
```

```python
lista = [90, 100, 1, 5, 10]

acum = 1

for i in lista:
    acum = acum * i

print(acum)
```

```python
lista = [2, 4, 5, 10, 15, 17]

acum = 0

for i in lista:
    if i % 2 == 0:
        acum = acum + 1

print(acum)
```

```python
list = [100, 400, 200, "", 300, "200", 120]

acum = 0

for i in list:
    if type(i) == str:
        continue
    acum = acum + i

print(acum)
```
