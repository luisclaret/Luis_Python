```python
try:
    number = 1
    string = "hola"
    result = string + string
    string[10]
except TypeError as error:
    print(error)
    print("Hay un error type error")
except IndexError as error:
    print(error)
    print("Hay un error de Index")
except:
    print("Hay un error generico")
```

```python
try:
    number = 1
    string = "hola"
    result = number + string
except Exception as error:
    print(error)
    print("Hay un error generico")
```

```python
try:
    10 / 0
except:
    print("Division por 0")
    raise ValueError("Division por 0 no se puede")
```

```python
comp = 1 > 0
comp2 = 0 > 0

if (comp):
    print("cuando comp es true")

if (comp):
    print("cuando comp es true")
else:
    print("cuando comp es false")


if (comp):
    print("cuando comp es true")
elif (comp and comp2):
    print("otra condicion pero negando la de arriba")
else: # OPCIONAL
    print("ninguna es true")
```

# FUNCIONES

```python
def nombre_funcion(a, b, c):
    return a + b, a + c, b + c, "hola" # En una tupla puedo guardar multiples tipos
```

```python
var = nombre_funcion(1, 2, 3) #el retorno de la funcion es una tupla
print(var) # Una tupla solo se puede crear e indexar
```

```python
print(var[0]) # El index en python comienza en 0
print(var[1])
print(var[2])
nombre_funcion(var[0], var[1], var[2])
```
